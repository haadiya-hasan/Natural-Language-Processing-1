# NLP Analysis - Tokenization

**Tokenization in NLP** is the process of breaking down text into smaller units called tokens, such as words, sentences, or characters. It is an important text preprocessing technique that helps computers process and analyze human language.

In Python, tokenization can be performed using libraries such as **NLTK** and **spaCy**. In this project, we will implement tokenization using both libraries one by one, exploring how each processes text through word and sentence tokenization and comparing their outputs.

## A. Tokenization with spaCy

### 1. Importing the Necessary Libraries

We need to install NLTK and spaCy for tokenization. We will also upload the extract.txt file, which contains the text that will be used for tokenization. After installing the libraries, we will load the required spaCy English language model.

```python
# Install NLTK and spaCy libraries
!pip install nltk spacy

# Download the small English language model for spaCy
!python -m spacy download en_core_web_sm


# Import the spaCy library
import spacy

# Load the English language model
nlp = spacy.load('en_core_web_sm')
```

### 2. Loading the Text File

```python
# Read the uploaded text file
with open("text.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Process the text using spaCy
doc = nlp(text)

print(text)
```

### 3. Basic Tokenization

```python
# Print each token from the uploaded text
for token in doc:
    print(token.text)
```

### 4. Tokenization with Token Index

```python
# Display each token with its position in the document
for token in doc:
    print(token.i, token.text)
```

### 5. Tokenization with Token Attributes

```python
# Display token, alphabetic, punctuation and space attributes
for token in doc:
    print(token.text, token.is_alpha, token.is_punct, token.is_space)
```

### 6. Tokenization with Punctuation

```python
# Display tokens containing punctuation
for token in doc:
    if token.is_punct:
        print(token.text)
```

### 8. Tokenization of Numbers

```python
# Identify and display numbers present in the text
for token in doc:
    if token.like_num:
        print(token.text)
```

### 9. Tokenization of Email Addresses

```python
# Check for email addresses in the uploaded text
for token in doc:
    if token.like_email:
        print(token.text)
```

### 10. Tokenization of URLs


```python
# Check for website URLs in the uploaded text
for token in doc:
    if token.like_url:
        print(token.text)
```

### 11. Tokenization of Contractions

```python
# Identify contractions in the uploaded text
for token in doc:
    if "'" in token.text or "’" in token.text:
        print(token.text)
```

### 12. Tokenization of Hyphenated Words

```python
# Identify hyphenated words from the uploaded text
for i, token in enumerate(doc):
    if token.text == "-" and i > 0 and i < len(doc) - 1:
        print(doc[i-1:i+2].text)
```

### 13. Removing Punctuation After Tokenization

```python
# Display tokens without punctuation
for token in doc:
    if not token.is_punct:
        print(token.text)
```

### 14. Getting Only Alphabetic Tokens

```python
# Display only tokens containing alphabetic characters
for token in doc:
    if token.is_alpha:
        print(token.text)
```

### 15. Converting Tokens into a List

```python
# Convert all tokens into a Python list
token_list = [token.text for token in doc]

print(token_list)
```

### 16. Tokenization of Multiple Sentences

```python
# Tokenize the complete passage
for token in doc:
    print(token.text)

# spaCy processes the entire passage as a Doc object
```

### 17. Tokenization and Sentence Segmentation

```python
# Identify individual sentences and their tokens
for sent in doc.sents:
    print("Sentence:", sent.text)

    for token in sent:
        print(token.text)

    print()
```

### 18. Complete Example

```python
# Display token attributes for the entire uploaded passage
print("Token\tAlpha\tNumber\tPunctuation")

for token in doc:
    print(
        token.text,
        "\t",
        token.is_alpha,
        "\t",
        token.like_num,
        "\t",
        token.is_punct
    )
```

## B. Tokenization with NLTK

### 1. Introduction

Tokenization is an important text preprocessing technique in Natural Language Processing (NLP). It is the process of breaking down text into smaller units called tokens, such as words and sentences.

In this practical, we will use the NLTK (Natural Language Toolkit) library in Python to perform different types of tokenization and understand how it handles punctuation, numbers, contractions, URLs, email addresses, and hyphenated words.

### 2. Importing NLTK

```python
# Import the NLTK library
import nltk
import string
import re

# Import required tokenization functions
from nltk.tokenize import word_tokenize, sent_tokenize, wordpunct_tokenize

# Download the required tokenization resource
nltk.download('punkt_tab')
```

### 3. Word Tokenization

```python
# Divide the complete text into individual words and punctuation
tokens = word_tokenize(text)

print(tokens)
```

### 4. Sentence Tokenization

```python
# Divide the text into individual sentences
sentences = sent_tokenize(text)

for sentence in sentences:
    print(sentence)
```

### 5. Word Tokenization of Multiple Sentences

```python
# Tokenize words from the entire text containing multiple sentences
tokens = word_tokenize(text)

print(tokens)
```

### 6. Tokenization with Punctuation

```python
# Display words and punctuation separately
tokens = word_tokenize(text)

for token in tokens:
    print(token)
```

### 7. Tokenization of Numbers

```python
# Identify and display numbers from the text
tokens = word_tokenize(text)

for token in tokens:
    if any(char.isdigit() for char in token):
        print(token)
```

### 8. Tokenization of Email Addresses

```python
# Identify email addresses in the text
import re

emails = re.findall(r'[\w.-]+@[\w.-]+\.\w+', text)

print(emails)
```

### 9. Tokenization of URLs

```python
# Identify URLs in the text
urls = re.findall(r'https?://\S+', text)

print(urls)
```

### 10. Tokenization of Contractions

```python
# Identify contractions in the text
tokens = word_tokenize(text)

for token in tokens:
    if token in ["n't", "'m", "'re", "'ve", "'ll", "'d", "'s"]:
        print(token)
```

### 11. Tokenization of Hyphenated Words

```python
# Identify hyphenated words in the text
hyphenated_words = re.findall(r'\b\w+(?:-\w+)+\b', text)

print(hyphenated_words)
```

### 12. Count Number of Tokens

```python
# Count the total number of tokens
tokens = word_tokenize(text)

print("Tokens:", tokens)
print("Number of tokens:", len(tokens))
```

### 13. Count Only Words

```python
# Remove punctuation and count only alphabetic words
tokens = word_tokenize(text)

words = [token for token in tokens if token.isalpha()]

print("Words:", words)
print("Number of words:", len(words))
```

### 14. Convert Tokens to Lowercase

```python
# Convert all tokens into lowercase
tokens = word_tokenize(text)

lower_tokens = [word.lower() for word in tokens]

print("Original Tokens:", tokens)
print("Lowercase Tokens:", lower_tokens)
```

### 15. Tokenization Using wordpunct_tokenize()

```python
# Tokenize the text using wordpunct_tokenize()
tokens = wordpunct_tokenize(text)

print("Tokens:", tokens)
```

### 16. Tokenization of Paragraph

```python
# Divide the uploaded paragraph into sentences
sentences = sent_tokenize(text)

print("Number of sentences:", len(sentences))

for i, sentence in enumerate(sentences, start=1):
    print(f"\nSentence {i}:")
    print(sentence)
```

### 17. Tokenization from a Text File

```python
# Read the uploaded text file
with open("text.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Perform word tokenization
tokens = word_tokenize(text)

print("Tokens:", tokens)
```

### 18. Complete Tokenization Program

```python
# Step 1: Sentence Tokenization
sentences = sent_tokenize(text)

print("Sentences:")
for sentence in sentences:
    print(sentence)

# Step 2: Word Tokenization
tokens = word_tokenize(text)

print("\nAll Tokens:")
print(tokens)

# Step 3: Remove Punctuation
words = [word for word in tokens if word.isalpha()]

print("\nTokens without Punctuation:")
print(words)

# Step 4: Convert to Lowercase
lower_words = [word.lower() for word in words]

print("\nLowercase Tokens:")
print(lower_words)

# Step 5: Count Tokens
print("\nTotal Tokens:", len(tokens))
print("Total Words:", len(words))
```
