## 1. Overview

Stemming is a text-preprocessing technique used in Natural Language Processing (NLP). It reduces related words to a common base form, known as a *stem*, by removing prefixes or suffixes.

For example, words such as `playing`, `played`, and `plays` may be reduced to the stem `play`. A stem is not always a valid dictionary word; the objective is to group related word forms, not to produce grammatically perfect words.

> **Important:** spaCy does not provide a built-in stemming algorithm. It primarily supports lemmatization. In the examples below, spaCy is used to tokenize text, while NLTK performs the stemming.

## 2. Stemming and lemmatization

Both techniques reduce related word forms, but they work differently.

| Aspect | Stemming | Lemmatization |
|---|---|---|
| Method | Removes word endings using rules | Uses vocabulary and linguistic analysis |
| Result | May not be a dictionary word | Usually a valid base word |
| Context | Generally does not consider context | Can consider part of speech and context |
| Example | `studies` → `studi` | `studies` → `study` |

**When to use stemming:** It can be useful when speed and broad word matching are more important than producing linguistically correct base forms, such as in some search and information-retrieval workflows.

## 4. Stemming with NLTK

### 4.1 Import and initialize PorterStemmer

`PorterStemmer` applies a sequence of stemming rules to reduce words to their stems. We create one instance and reuse it throughout this section.

```python
# Import the PorterStemmer class from NLTK.
from nltk.stem import PorterStemmer

# Create a stemmer instance for the examples below.
stemmer = PorterStemmer()
```

### 4.2 Basic stemming

Start with a small group of words that share a common root. The loop applies the stemmer to each word and displays the original word alongside its stem.

```python
# Example 1: Apply stemming to individual word forms.
words = ["playing", "played", "plays", "player"]

for word in words:
    print(f"{word:<12} -> {stemmer.stem(word)}")
```

### 4.3 Stemming words ending in `-ing`

The following examples demonstrate how the stemmer processes common present-participle forms.

```python
# Example 2: Examine words ending in "-ing".
words = ["playing", "reading", "writing", "walking", "running"]

for word in words:
    print(f"{word:<12} -> {stemmer.stem(word)}")
```

### 4.4 Stemming words ending in `-ed`

Past-tense forms can also be reduced to a shared stem.

```python
# Example 3: Examine words ending in "-ed".
words = ["played", "walked", "talked", "connected", "completed"]

for word in words:
    print(f"{word:<12} -> {stemmer.stem(word)}")
```

### 4.5 Stemming plural words

This example checks how plural nouns are processed.

```python
# Example 4: Apply stemming to plural nouns.
words = ["cats", "dogs", "books", "cars", "students"]

for word in words:
    print(f"{word:<12} -> {stemmer.stem(word)}")
```

### 4.6 Comparing related word forms

Words that look related do not always produce the same stem. Stemming is rule-based, so the output should be interpreted as a text-processing result rather than a dictionary definition.

```python
# Example 5: Compare several forms related to "connect".
words = [
    "connect",
    "connected",
    "connecting",
    "connection",
    "connections"
]

for word in words:
    print(f"{word:<15} -> {stemmer.stem(word)}")
```

### 4.7 Comparing adjective and noun forms

The stemmer can also process comparative, superlative, and noun forms.

```python
# Example 6: Examine different forms related to "happy".
words = ["happy", "happier", "happiest", "happiness"]

for word in words:
    print(f"{word:<12} -> {stemmer.stem(word)}")
```

### 4.8 Stemming words related to "organize"

This example highlights how a stemmer may reduce several related forms to a common stem.

```python
# Example 7: Compare forms related to "organize".
words = [
    "organize",
    "organized",
    "organizing",
    "organization",
    "organizations"
]

for word in words:
    print(f"{word:<16} -> {stemmer.stem(word)}")
```

### 4.9 Stemming a sentence

A sentence can be split into individual words and each word can then be stemmed. This simple approach is useful for demonstration, although `split()` does not remove punctuation.

```python
# Example 8: Split a sentence into words and stem each word.
text = "The students are studying different subjects."

words = text.lower().split()

for word in words:
    print(f"{word:<14} -> {stemmer.stem(word)}")
```

### 4.10 Tokenizing text with a regular expression

Regular expressions provide more control over token extraction. The pattern below extracts alphabetic sequences and excludes punctuation before stemming.

```python
# Example 9: Extract alphabetic tokens, convert them to lowercase,
# and apply stemming to each token.

import re

text = (
    "Students are learning Python and developing "
    "machine learning projects."
)

words = re.findall(r"\b[a-zA-Z]+\b", text.lower())

for word in words:
    print(f"{word:<14} -> {stemmer.stem(word)}")
```

### 4.11 Reconstructing a stemmed sentence

Instead of printing each stem separately, store the results in a list and join them to form a stemmed version of the original sentence.

```python
# Example 10: Build a new sentence from the stemmed words.
text = "The players are playing football and practicing regularly."

# Convert to lowercase and split the sentence into tokens.
words = text.lower().split()

# Store each stem in a separate list.
stemmed_words = [stemmer.stem(word) for word in words]

# Join the processed tokens into one string.
stemmed_text = " ".join(stemmed_words)

print("Original sentence:")
print(text)

print("\nStemmed sentence:")
print(stemmed_text)
```

## 5. Using spaCy for tokenization and NLTK for stemming

spaCy can identify tokens in a text, including words and punctuation. Since spaCy does not include a stemming algorithm, we pass each alphabetic token to NLTK's `PorterStemmer`.

This combination demonstrates how two NLP libraries can be used together, with each library handling a specific task.

```python
# Import spaCy and the NLTK stemmer.
import spacy
from nltk.stem import PorterStemmer

# Load spaCy's English language model.
nlp = spacy.load("en_core_web_sm")

# Initialize the NLTK stemmer.
stemmer = PorterStemmer()
```

### 5.1 Tokenization followed by stemming

First, spaCy divides the sentence into tokens. The `is_alpha` condition keeps alphabetic tokens and skips punctuation and numeric tokens.

```python
# Example 1: Tokenize a sentence with spaCy and stem its words with NLTK.
text = "The students are studying."

doc = nlp(text)

for token in doc:
    if token.is_alpha:
        print(f"{token.text:<12} -> {stemmer.stem(token.text.lower())}")
```

### 5.2 Playing-related words

Observe how different grammatical forms of `play` are processed.

```python
# Example 2: Process playing, played, and plays in context.
text = "I play football. He plays football. They played yesterday."

doc = nlp(text)

for token in doc:
    if token.is_alpha:
        print(f"{token.text:<12} -> {stemmer.stem(token.text.lower())}")
```

### 5.3 Reading-related words

The following sentence contains several forms related to reading. Each alphabetic token is stemmed independently.

```python
# Example 3: Process words related to reading.
text = "Reading readers read readable books."

doc = nlp(text)

for token in doc:
    if token.is_alpha:
        print(f"{token.text:<12} -> {stemmer.stem(token.text.lower())}")
```

### 5.4 Learning-related words

This example compares the base form, present participle, and past-tense form of `learn`.

```python
# Example 4: Process different forms of "learn".
text = "Students learn, learning, and learned concepts."

doc = nlp(text)

for token in doc:
    if token.is_alpha:
        print(f"{token.text:<12} -> {stemmer.stem(token.text.lower())}")
```

### 5.5 Processing a complete paragraph

A paragraph can be processed in the same way as a sentence. spaCy handles tokenization across the complete passage, and NLTK generates a stem for each alphabetic token.

The passage below is a short extract from **_Alice's Adventures in Wonderland_ by Lewis Carroll (1865)**, a public-domain work.

```python
# Example 5: Apply stemming to a literary text extract.
text = (
    "There was nothing so VERY remarkable in that; nor did Alice think it "
    "so VERY much out of the way to hear the Rabbit say to itself, "
    "'Oh dear! Oh dear! I shall be late!' (when she thought it over "
    "afterwards, it occurred to her that she ought to have wondered at "
    "this, but at the time it all seemed quite natural); but when the "
    "Rabbit actually TOOK A WATCH OUT OF ITS WAISTCOAT-POCKET, and looked "
    "at it, and then hurried on, Alice started to her feet, for it flashed "
    "across her mind that she had never before seen a rabbit with either "
    "a waistcoat-pocket, or a watch to take out of it, and burning with "
    "curiosity, she ran across the field after it."
)

doc = nlp(text)

print("TOKEN".ljust(16), "STEM")
print("-" * 25)

for token in doc:
    if token.is_alpha:
        print(f"{token.text.lower():<16} {stemmer.stem(token.text.lower())}")
```

### 5.6 Handling punctuation

spaCy identifies punctuation as separate tokens. The `is_alpha` attribute allows us to skip punctuation without manually removing it from the original text.

```python
# Example 6: Keep alphabetic tokens and ignore punctuation.
text = "Python is powerful, useful, and popular!"

doc = nlp(text)

for token in doc:
    if token.is_alpha:
        print(f"{token.text:<12} -> {stemmer.stem(token.text.lower())}")
```

### 5.7 Normalizing uppercase and lowercase text

Stemming is case-sensitive in its input handling, so converting tokens to lowercase first provides consistent processing for words that differ only by capitalization.

```python
# Example 7: Convert each token to lowercase before stemming.
text = "Python Pythonic Programming Programs"

doc = nlp(text)

for token in doc:
    if token.is_alpha:
        normalized_word = token.text.lower()
        print(f"{token.text:<16} -> {stemmer.stem(normalized_word)}")
```

### 5.8 Reconstructing a stemmed sentence with spaCy

This final example collects the stems of alphabetic tokens and joins them into a new sentence. Punctuation is intentionally excluded to keep the reconstruction straightforward.

```python
# Example 8: Create a stemmed sentence using spaCy tokenization.
text = "The children are playing games and enjoying themselves."

doc = nlp(text)

# Collect the stem of each alphabetic token.
stemmed_words = []

for token in doc:
    if token.is_alpha:
        stemmed_words.append(stemmer.stem(token.text.lower()))

# Reconstruct the sentence from the processed tokens.
stemmed_text = " ".join(stemmed_words)

print("Original sentence:")
print(text)

print("\nStemmed sentence:")
print(stemmed_text)
```

## 6. Key observations

- Stemming reduces related word forms using predefined rules.
- The resulting stem may not be a valid English word.
- NLTK's `PorterStemmer` can process individual words, sentences, and longer passages.
- Regular expressions and spaCy can help identify tokens before stemming.
- spaCy performs tokenization in the combined examples; NLTK performs the stemming.
- Lowercasing before stemming helps maintain consistent treatment of differently capitalized words.

## 7. Conclusion

Stemming is a useful and relatively simple NLP preprocessing technique for reducing word variation. This notebook demonstrated stemming with NLTK's `PorterStemmer`, text token extraction with regular expressions, and a combined spaCy–NLTK workflow. The examples also show why stemming should be understood as a rule-based normalization step rather than a method for finding the exact dictionary form of every word.
