## 2. Importing the Necessary Libraries

```python
# Import the library used to read and write PDF files.
from PyPDF2 import PdfReader, PdfWriter

# Import spaCy for Natural Language Processing.
import spacy

# Import Matcher to identify specific patterns in text.
from spacy.matcher import Matcher

# Import Counter to count the frequency of words and entities.
from collections import Counter

# Load spaCy's English language model.
nlp = spacy.load("en_core_web_sm")
```

## 3. Read the PDF File

### **Exercise 1** Reading a PDF

In this exercise, we will open `Uncle Tungsten.pdf` using the **PdfReader** class from `PyPDF2`. We will also determine the total number of pages in the document.

This is the first step because the PDF's contents must be extracted before we can analyze the text.

```python
# Open the PDF file using PyPDF2.
reader = PdfReader("Uncle Tungsten.pdf")

# Display the total number of pages in the PDF.
print("Total number of pages:", len(reader.pages))
```

## 4. Extract text from the first page

### **Exercise 2:** Extract Text from the First Page
A PDF stores its content in a document format. We can extract the text from a particular page using the `extract_text()` method.

Here, we extract and display the text from the first page of The Big Short.pdf.

```python
# Select the first page of the PDF.
first_page = reader.pages[0]

# Extract the text from the selected page.
first_page_text = first_page.extract_text()

# Display the extracted text.
print(first_page_text)
```

## 5. Extract text from all Pages

###**Exercise 3:** Extract Text from the Entire PDF
In this exercise, we will extract text from every page of the PDF. The extracted text from each page will be stored in a list called `page_texts`.

We will then combine the text from all pages into one string called full_text. This combined text will be used for the NLP exercises later in the notebook.

```python
# Create an empty list to store the text extracted from each page.
page_texts = []

# Visit each page and extract its text.
for page_number, page in enumerate(reader.pages, start=1):

    # Extract text from the current page.
    page_text = page.extract_text()

    # Store an empty string if the page has no extractable text.
    if page_text is None:
        page_text = ""

    # Add the extracted text to the list.
    page_texts.append(page_text)

# Combine all page texts into one string.
full_text = "\n".join(page_texts)

# Display the number of pages processed.
print("Pages processed:", len(page_texts))

# Display the total number of characters extracted.
print("Total characters extracted:", len(full_text))

# Display the beginning of the extracted text.
print(full_text[:2000])
```

**Observation:**

The extracted text is stored in full_text. The first 2,000 characters are displayed to verify that the PDF was read successfully.

Some PDFs contain scanned images rather than selectable text. If the extracted text is empty or incomplete, the document may require Optical Character Recognition (OCR).

## 6. Extract text from a specific page

###**Exercise 4:** Read a Specific Page
Sometimes, we need to read only one page instead of processing the entire document. We can select a page by its index.

Python uses zero-based indexing, so page index `0` represents page `1`, page index `1` represents page `2`, and so on.

```python
# Specify the page number we want to read.
# This example reads page 5.
page_number = 5

# Check that the requested page exists in the PDF.
if 1 <= page_number <= len(reader.pages):

    # Extract text using the zero-based page index.
    selected_page_text = reader.pages[page_number - 1].extract_text()

    # Display the selected page's text.
    print(selected_page_text)

else:
    # Display an error message if the page number is invalid.
    print("The requested page number does not exist.")
```

## 7. Search for a word or phrase

###**Exercise 5:** Search for a Word or Phrase
Searching for specific words can help us locate relevant information in a long document.

In this exercise, we will search for the word mortgage in every page of the PDF. The search is case-insensitive, which means it treats mortgage, Mortgage, and MORTGAGE as the same word.

```python
# Specify the word or phrase to search for.
search_word = "Tungsten"

# Create a list to store the page numbers containing the search term.
matching_pages = []

# Search through every page of the PDF.
for page_number, page_text in enumerate(page_texts, start=1):

    # Convert both strings to lowercase for a case-insensitive search.
    if search_word.lower() in page_text.lower():

        # Store the page number if the search term is found.
        matching_pages.append(page_number)

        # Display the page number containing the search term.
        print("Found on page:", page_number)

# Display a message if the search term does not occur.
if not matching_pages:
    print("The search term was not found in the PDF.")
```

## 8. Count the words in the PDF

###**Exercise 6:** Count Words in the Document
Word counting is a basic text-processing task. In this exercise, we will count the words extracted from every page of the PDF.

The `split()` method separates text into whitespace-delimited pieces. This gives us a simple word-count estimate, although punctuation and unusual formatting may affect the result slightly.

```python
# Initialize the total word count.
total_words = 0

# Count the words extracted from each page.
for page_text in page_texts:

    # Split the page text into words and add the count.
    total_words += len(page_text.split())

# Display the total number of words in the document.
print("Total word count:", total_words)
```

## 9. Save the first page as a new PDF

### **Exercise 7:** Copy the First Page into a new PDF.
`PyPDF2` can copy existing PDF pages into a new PDF document. In this exercise, we will create a separate PDF containing only the first page of `Uncle Tungsten.pdf`

This preserves the original page instead of recreating its layout.

```python
# Create a PDF writer object for the new PDF.
writer = PdfWriter()

# Add the first page of the original PDF.
writer.add_page(reader.pages[0])

# Open a new PDF file in binary write mode.
with open("The_Big_Short_First_Page.pdf", "wb") as output_file:

    # Write the selected page to the new PDF.
    writer.write(output_file)

# Confirm that the new PDF has been created.
print("First page saved as The_Big_Short_First_Page.pdf")
```

## 10. Save the extracted text as a TXT file

###**Exercise 8:** Save Extracted Text to a Text File
After extracting text from the PDF, we can save it in a plain-text file. A TXT file is useful because its contents can be opened, searched, edited, and reused without needing a PDF reader.

In this exercise, we will save the extracted text and label each page so that the page boundaries remain identifiable.

```python
# Open a text file in write mode using UTF-8 encoding.
with open("The_Big_Short_Extracted_Text.txt", "w", encoding="utf-8") as output_file:

    # Write the text from each page to the output file.
    for page_number, page_text in enumerate(page_texts, start=1):

        # Add a heading before the text from each page.
        output_file.write(f"\n--- Page {page_number} ---\n")

        # Write the extracted text of the current page.
        output_file.write(page_text)

# Confirm that the text file has been created.
print("Extracted text saved as The_Big_Short_Extracted_Text.txt")
```

## 11. Prepare the text for NLP

###**Exercise 9: Set Up spaCy**
We will now use spaCy to analyze the extracted text. The English language model provides tokenization, part-of-speech tags, lemmas, grammatical information, and named entities.

A token is a unit of text, such as a word or punctuation mark. A lemma is the base form of a word. For example, the lemma of running is usually run.

```python
# Import spaCy for natural language processing.
import spacy

# Import Matcher to find patterns in the text.
from spacy.matcher import Matcher

# Import displaCy to visualize named entities.
from spacy import displacy

# Import Counter to count repeated items.
from collections import Counter

# Load the small English language model.
nlp = spacy.load("en_core_web_sm")

# Increase the maximum allowed text length if the PDF is very long.
nlp.max_length = max(nlp.max_length, len(full_text) + 100)

# Process the complete extracted text with spaCy.
doc = nlp(full_text)

# Display basic information about the NLP document.
print("spaCy version:", spacy.__version__)
print("Total tokens:", len(doc))
print("Total sentences:", len(list(doc.sents)))
```

## 12. Identify stopwords

### **Exercise 10: Identify Stopwords**
Stopwords are common words that may contribute less meaning in certain text-analysis tasks. Examples include `the`, `is`, `and`, `of` and `to`.

**spaCy** provides the is_stop attribute, which identifies whether a token is considered a stopword by its language model.

In this exercise, we will identify the stopwords present in the PDF and count how frequently they occur.

```python
# Select alphabetic tokens that spaCy identifies as stopwords.
stopword_list = [
    token.text.lower()
    for token in doc
    if token.is_alpha and token.is_stop
]

# Count the occurrences of each stopword.
stopword_counts = Counter(stopword_list)

# Display the 30 most frequently occurring stopwords.
print("Most frequent stopwords:\n")

for word, count in stopword_counts.most_common(30):

    # Print each stopword and its frequency.
    print(word, ":", count)

# Display the total number of stopword occurrences.
print("\nTotal stopword occurrences:", len(stopword_list))

# Display the number of unique stopwords.
print("Unique stopwords:", len(stopword_counts))
```

**Observation**

The output shows common words found throughout the document and how often they occur. Stopwords are not always meaningless; whether they should be removed depends on the NLP task being performed.

## 13. Use Matcher to identify patterns

### **Exercise 11: Pattern Matching with spaCy Matcher**
The Matcher class allows us to identify words and phrases that follow specific linguistic patterns.

Unlike a simple keyword search, Matcher can use information such as a token's lowercase form, part of speech, or grammatical characteristics.

In this exercise, we will search for patterns related to finance and the financial problems discussed in the document. The patterns are examples and can be modified to find other expressions.

```python
# Create a Matcher object using the vocabulary of the NLP model.
matcher = Matcher(nlp.vocab)

# Define a pattern that matches the word "tungsten".
tungsten_pattern = [
    {"LOWER": "tungsten"}
]

# Define a pattern that matches "periodic table".
periodic_table_pattern = [
    {"LOWER": "periodic"},
    {"LOWER": "table"}
]

# Define a pattern that matches "light bulb" or "light bulbs".
light_bulb_pattern = [
    {"LOWER": "light"},
    {"LOWER": {"IN": ["bulb", "bulbs"]}}
]

# Define a pattern that matches "Uncle Dave".
uncle_dave_pattern = [
    {"LOWER": "uncle"},
    {"LOWER": "dave"}
]

# Define a pattern that matches "chemical" followed by any noun (chemical reaction, chemical element...).
chemical_term_pattern = [
    {"LOWER": "chemical"},
    {"POS": "NOUN"}
]

# Define a pattern that matches names of metals.
metal_pattern = [
    {"LOWER": {"IN": ["gold", "silver", "iron", "copper", "lead", "zinc", "mercury", "platinum", "osmium"]}}
]

# Register the patterns with names so their matches can be identified.
matcher.add("TUNGSTEN", [tungsten_pattern])
matcher.add("PERIODIC_TABLE", [periodic_table_pattern])
matcher.add("LIGHT_BULB", [light_bulb_pattern])
matcher.add("UNCLE_DAVE", [uncle_dave_pattern])
matcher.add("CHEMICAL_TERM", [chemical_term_pattern])
matcher.add("METAL", [metal_pattern])

# Run the matcher on the processed document.
matches = matcher(doc)

# Display every matched phrase.
print("Matched patterns:\n")

for match_id, start, end in matches:

    # Convert the numeric match ID into its readable pattern name.
    pattern_name = nlp.vocab.strings[match_id]

    # Extract the text corresponding to the matched token range.
    matched_text = doc[start:end].text

    # Display the pattern name and the matching text.
    print(pattern_name, ":", matched_text)

# Display the total number of pattern matches.
print("\nTotal pattern matches:", len(matches))
```

**Observation**

The Matcher only finds phrases that follow the registered token pattern exactly. A phrase is matched only if its tokens appear in the same order and match each token's rule. For example, the two-token periodic table pattern will not match "periodic tables", because the second token is "tables", not "table". The light bulb pattern will also miss "light-bulb" or "lightbulb". spaCy splits "light-bulb" into three tokens (light, -, bulb), so the pattern, which expects two tokens, finds nothing. The tungsten pattern matches only the single word "tungsten", so a variant like "tungstens" or a related word such as "wolfram" will not match. The Matcher does not understand meaning; it only matches the patterns it was given.

## 14. Perform lemmatization

###**Exercise 12: Lemmatization**
Lemmatization converts a word into its dictionary base form, known as its lemma. For example, running may be reduced to run, banks to bank, and went to go.

In this exercise, we will display the original words and their lemmas. We will focus on alphabetic tokens to keep the output readable.

```python
# Create a list containing each token and its lemma.
lemmatization_results = []

# Process each token in the document.
for token in doc:

    # Ignore punctuation, numbers, and other non-alphabetic tokens.
    if token.is_alpha:

        # Store the original word and its base form.
        lemmatization_results.append(
            (token.text, token.lemma_)
        )

# Display the first 100 word and lemma pairs.
print("Original word -> Lemma\n")

for word, lemma in lemmatization_results[:100]:

    # Display the original word alongside its lemma.
    print(word, "->", lemma)

# Display the number of alphabetic tokens processed.
print("\nAlphabetic tokens processed:", len(lemmatization_results))
```

**Observation**

Lemmatization helps normalize different forms of the same word. It is useful in tasks such as document search, text classification, and information retrieval because words with related grammatical forms can be grouped together.

## 15. Find nouns whose forms change after lemmatization

###**Exercise 13: Identify Nouns That Change After Lemmatization**
Nouns may appear in singular or plural forms. For example, `banks` may have the lemma bank, while `companies` may have the lemma company.

In this exercise, we will identify nouns whose original text differs from the lemma assigned by spaCy. The results may include plural nouns and other noun forms that are normalized by the language model.

```python
# Create an empty list to store nouns whose forms change.
changed_nouns = []

# Examine each token in the document.
for token in doc:

    # Select tokens tagged as nouns.
    if token.pos_ == "NOUN":

        # Compare the lowercase original form with the lowercase lemma.
        if token.text.lower() != token.lemma_.lower():

            # Store the original word, lemma, and grammatical tag.
            changed_nouns.append(
                (token.text, token.lemma_, token.tag_)
            )

# Remove duplicate word-lemma pairs while preserving their order.
unique_changed_nouns = []
seen_nouns = set()

for word, lemma, tag in changed_nouns:

    # Use lowercase forms to avoid duplicates caused only by capitalization.
    key = (word.lower(), lemma.lower())

    # Add the pair only if it has not already been recorded.
    if key not in seen_nouns:
        seen_nouns.add(key)
        unique_changed_nouns.append((word, lemma, tag))

# Display the first 100 unique noun forms that changed.
print("Nouns that change after lemmatization:\n")

for word, lemma, tag in unique_changed_nouns[:100]:

    # Display the original noun, its lemma, and its grammatical tag.
    print(f"{word} -> {lemma} | Tag: {tag}")

# Display the number of unique changed noun forms.
print("\nUnique changed noun forms:", len(unique_changed_nouns))
```

**Observation**

This exercise shows how spaCy normalizes noun forms. The output depends on the vocabulary and grammar present in `Uncle Tungsten.pdf`. Not every noun will change, because many nouns are already in their base form.

## 16. Identify auxiliary verbs

###**Exercise 14: Find Auxiliary Verbs**
Auxiliary verbs, also known as helping verbs, work with main verbs to express tense, voice, possibility, necessity, and other grammatical meanings.

Examples include `is`, `are`, `was`, `were`, `has`, `have`, `had`, `will`, `can`, and `should`.

spaCy labels auxiliary verbs with the part-of-speech tag AUX. In this exercise, we will extract and count the auxiliary verbs found in the PDF.

```python
# Select all tokens that spaCy identifies as auxiliary verbs.
auxiliary_verbs = [
    token.text
    for token in doc
    if token.pos_ == "AUX"
]

# Count the frequency of each auxiliary verb.
auxiliary_verb_counts = Counter(
    verb.lower() for verb in auxiliary_verbs
)

# Display the most frequent auxiliary verbs.
print("Auxiliary verb frequencies:\n")

for verb, count in auxiliary_verb_counts.most_common():

    # Print each auxiliary verb and its frequency.
    print(verb, ":", count)

# Display the total number of auxiliary verb occurrences.
print("\nTotal auxiliary verb occurrences:", len(auxiliary_verbs))
```

Observation
The frequency counts show which auxiliary verbs occur most often in the extracted text. The classification is based on spaCy's grammatical analysis, so some words may be tagged differently depending on how they are used in a sentence.

## 17. Find adjectives describing data-related problems

###**Exercise 15: Identify Adjectives Associated with Data Problems**
Adjectives describe or qualify nouns. In a chemical text, adjectives may be used to describe reactions, compounds, substances, properties, hazards, and the quality of experimental data.

For this exercise, we will first identify adjectives appearing near selected terms related to chemistry problems and substances. We will also retain the surrounding sentence so that the adjectives can be interpreted in context.

The selected keywords are examples, not a definitive list of every possible data problem.

```python
# Define keywords associated with chemical hazards, experiments and laboratory problems.
problem_keywords = {
    "experiment", "experiments",
    "reaction", "reactions",
    "explosion", "explosions",
    "danger", "dangers",
    "accident", "accidents",
    "poison", "poisons",
    "fumes", "fume",
    "acid", "acids",
    "impurity", "impurities",
    "error", "errors",
    "failure", "failures",
    "waste",
    "smell", "smells",
    "corrosion",
    "metal", "metals",
    "tungsten",
    "chemical", "chemicals",
    "element", "elements",
    "sample", "samples"
}

# Create a list to store adjectives found near the selected keywords.
problem_adjectives = []

# Examine every sentence in the document.
for sentence in doc.sents:

    # Examine each token in the current sentence.
    for index, token in enumerate(sentence):

        # Continue only if the token is an adjective.
        if token.pos_ == "ADJ":

            # Define a small window around the adjective.
            start = max(0, index - 4)
            end = min(len(sentence), index + 5)

            # Extract the words surrounding the adjective.
            context_tokens = list(sentence)[start:end]

            # Convert the context to lowercase for keyword checking.
            context_words = {
                item.text.lower()
                for item in context_tokens
            }

            # Check whether a selected problem-related keyword is nearby.
            if context_words.intersection(problem_keywords):

                # Save the adjective, its lemma, and the complete sentence.
                problem_adjectives.append(
                    (
                        token.text,
                        token.lemma_,
                        sentence.text.strip()
                    )
                )

# Remove duplicate adjective and context combinations.
unique_problem_adjectives = []
seen_adjectives = set()

for adjective, lemma, context in problem_adjectives:

    # Create a key to identify duplicate results.
    key = (adjective.lower(), context.lower())

    # Keep each unique result once.
    if key not in seen_adjectives:
        seen_adjectives.add(key)
        unique_problem_adjectives.append(
            (adjective, lemma, context)
        )

# Display the first 100 results with their contexts.
print("Adjectives found near chemistry-problem keywords:\n")

for adjective, lemma, context in unique_problem_adjectives[:100]:

    # Display the adjective, its lemma, and the sentence where it occurs.
    print("Adjective:", adjective)
    print("Lemma:", lemma)
    print("Context:", context)
    print("-" * 80)

# Display the number of unique results.
print("Unique adjective-context results:", len(unique_problem_adjectives))
```

**Observation**

This method identifies adjectives appearing near selected problem-related terms. It does not prove that every adjective describes a data problem; the surrounding sentence must be read to determine its actual meaning.

## 18. Recognize dates and monetary values

###**Exercise 16: Identify Dates and Money**
Named Entity Recognition (NER) is an NLP technique that identifies real-world entities mentioned in text.

spaCy uses the label DATE for date-related expressions and MONEY for monetary values. For example, a sentence might contain a date such as March 2008 or a monetary expression such as $5 million.

In this exercise, we will extract all entities labeled DATE or MONEY from The `Uncle Tungsten.pdf`.

```python
# Create an empty list to store dates and monetary values.
date_money_entities = []

# Examine all named entities identified by spaCy.
for entity in doc.ents:

    # Select entities labeled as dates or monetary values.
    if entity.label_ in {"DATE", "MONEY"}:

        # Store the entity text and its label.
        date_money_entities.append(
            (entity.text, entity.label_)
        )

# Display the extracted dates and monetary values.
print("Dates and monetary values:\n")

for entity_text, entity_label in date_money_entities:

    # Print the entity type and the text recognized by spaCy.
    print(f"{entity_label}: {entity_text}")

# Display the total number of matching entities.
print("\nTotal DATE and MONEY entities:", len(date_money_entities))
```

**Observation**

The output contains date expressions and monetary values that spaCy recognized in the PDF. Recognition is not guaranteed to be perfect, especially when the PDF has unusual formatting or when amounts are expressed in an unfamiliar way.

## 19. Recognize people, organizations, and places

###**Exercise 17: Identify People, Organizations, and Places**
Named Entity Recognition can also identify names of people, companies, institutions, and geographical locations.

spaCy uses several labels for these entities:
- `PERSON`: names of people.
- `ORG`: companies, agencies, institutions, and other organizations.
- `GPE`: countries, cities, states, and other geopolitical entities.
- `LOC`: geographical locations that are not geopolitical entities.
- `FAC`: facilities such as buildings, airports, and bridges.
In this exercise, we will extract entities belonging to these categories from the PDF.

```python
# Define the entity labels we want to identify.
target_entity_labels = {
    "PERSON",
    "ORG",
    "GPE",
    "LOC",
    "FAC"
}

# Create an empty list to store the selected entities.
named_entities = []

# Examine every entity recognized by spaCy.
for entity in doc.ents:

    # Keep only entities belonging to the selected categories.
    if entity.label_ in target_entity_labels:

        # Store the entity text and its label.
        named_entities.append(
            (entity.text, entity.label_)
        )

# Display the identified people, organizations, and places.
print("People, organizations, and places:\n")

for entity_text, entity_label in named_entities:

    # Print each entity with its category.
    print(f"{entity_label}: {entity_text}")

# Display the total number of selected entity occurrences.
print("\nTotal entity occurrences:", len(named_entities))
```

**Observation**

The extracted entities can help identify important people, financial institutions, and locations mentioned in the document. The results are generated by spaCy's trained model, so some entities may be missed or assigned an incorrect category.

## 20. Perform a complete Named Entity Recognition analysis

###**Exercise 18: Named Entity Recognition (NER)**
Named Entity Recognition is the process of locating and classifying named entities in text. These entities can include `people`, `organizations`, `places`, `dates`, `monetary values`, `percentages`, `products`, and other categories.

In this exercise, we will count all entity categories detected in the document and display the unique entities associated with each category.

```python
# Count how many times each entity category occurs.
entity_counts = Counter(
    entity.label_ for entity in doc.ents
)

# Display the total number of entities in each category.
print("Entity counts by category:\n")

for label, count in entity_counts.most_common():

    # Display the category and its occurrence count.
    print(label, ":", count)

# Create a dictionary to group unique entity names by category.
entities_by_type = {}

# Examine all entities identified by spaCy.
for entity in doc.ents:

    # Create a set for the category if it does not already exist.
    if entity.label_ not in entities_by_type:
        entities_by_type[entity.label_] = set()

    # Add the entity text to the appropriate category.
    entities_by_type[entity.label_].add(entity.text)

# Display the unique entities in each category.
print("\nUnique entities by category:\n")

for label in sorted(entities_by_type):

    # Display the name of the entity category.
    print("\nCategory:", label)

    # Sort entity names to make the output easier to read.
    for entity_name in sorted(entities_by_type[label], key=str.lower):

        # Display each unique entity name.
        print("-", entity_name)
```

**Observation**

The entity counts provide a summary of the types of real-world information detected in the document. Grouping unique entities makes it easier to inspect the people, companies, dates, locations, and monetary values mentioned throughout the text.
