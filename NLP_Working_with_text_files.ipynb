## 2. Creating and Writing to a Text File

Python's `open()` function is used to open a file.

The `"w"` mode opens a file for writing. If the file does not exist, Python creates it. If it already exists, its previous contents are overwritten.

The `write()` method writes a string into the file.


```python
# Create a text file and write content into it

file = open("example.txt", "w")

file.write("A Christmas Carol\n")
file.write("Written by Charles Dickens\n")
file.write("The main character is Ebenezer Scrooge.\n")

# Close the file after writing
file.close()

print("File created successfully.")
```

## 3. Reading the Contents of a File

The `"r"` mode opens a file for reading.

The `read()` method reads the contents of the file as a string. Once the cursor reaches the end of the file, another call to `read()` returns an empty string unless the cursor is moved back.


```python
# Open the file in read mode

my_file = open("example.txt", "r")

# Read and display the complete contents
print(my_file.read())

# Close the file
my_file.close()
```

## 4. Understanding the File Cursor

A file cursor keeps track of the current reading or writing position.

- `read()` moves the cursor forward.
- Reading again at the end of a file returns an empty string.
- `seek(0)` moves the cursor back to the beginning.

This helps us read the same file more than once without reopening it.


```python
# Open the file in read mode
my_file = open("example.txt", "r")

# Read the entire file
print(my_file.read())

# The cursor is now at the end
print(my_file.read())

# Move the cursor back to the beginning
my_file.seek(0)

# Read the file again
print(my_file.read())

# Close the file
my_file.close()
```

## 5. Reading Individual Lines

Python provides two useful methods for reading lines:

- `readline()` reads one line at a time.
- `readlines()` reads all lines and returns them as a list.

The newline character `\n` may remain at the end of each line.


```python
# Read the first two lines of the file

with open("example.txt", "r") as file:
    line1 = file.readline()
    line2 = file.readline()

print("First line:", line1.strip())
print("Second line:", line2.strip())
```

```python
# Read all lines into a list

with open("example.txt", "r") as file:
    lines = file.readlines()

print(lines)
```

## 6. Writing and Appending to Files

Python offers different file modes depending on the operation.

- `"w"` writes content and replaces existing contents.
- `"a"` adds new content at the end of a file.
- `"w+"` allows reading and writing but truncates existing contents.
- `"a+"` allows reading and appending.

**Important:** We will use a separate demonstration file for these operations so that the original book is not overwritten.


```python
# Write and read using w+ mode

my_file = open("christmas_carol_demo.txt", "w+")

my_file.write("Scrooge was a miserly man.\n")
my_file.write("Christmas changed his life.")

# Move the cursor to the beginning before reading
my_file.seek(0)

print(my_file.read())

my_file.close()
```

```python
# Append new content using a+ mode

my_file = open("christmas_carol_demo.txt", "a+")

# Add content at the end of the file
my_file.write("\nScrooge learned the importance of kindness.")

# Move the cursor to the beginning to read everything
my_file.seek(0)

print(my_file.read())

my_file.close()
```

## 7. Using `with open()` and `writelines()`

The `with open()` statement automatically closes the file after the block executes.

The `writelines()` method writes multiple strings from a list or another iterable. It does **not** add newline characters automatically, so we include `\n` in the strings where required.


```python
# Store multiple chapter titles in a list

chapters = [
    "Stave One: Marley's Ghost\n",
    "Stave Two: The First of the Three Spirits\n",
    "Stave Three: The Second of the Three Spirits\n",
    "Stave Four: The Last of the Spirits\n",
    "Stave Five: The End of It\n"
]

# Write all the chapter titles to a file
with open("christmas_carol_chapters.txt", "w") as file:
    file.writelines(chapters)

print("Chapter titles saved successfully.")
```

## 8. Reading *A Christmas Carol*

We will now work with the uploaded book.

The `encoding="utf-8-sig"` argument supports UTF-8 text files that may contain a byte-order mark at the beginning. It also reads ordinary UTF-8 text files.

We will use `with open()` so that the file is automatically closed when reading is complete.


```python
# Store the filename in a variable

book_file = "a_christmas_carol.txt"

# Read the entire book
with open(book_file, "r", encoding="utf-8-sig") as file:
    data = file.read()

# Display the book's contents
print(data)
```

## 9. Counting Lines, Words, and Characters

Python can perform basic text analysis using built-in functions.

- `splitlines()` separates text into lines.
- `split()` separates text into whitespace-delimited words.
- `len()` counts the items in a string or list.

These counts are based on the actual text file, including any title pages, introductory material, or other content it contains.


```python
# Count the lines, words, and characters in the book

with open(book_file, "r", encoding="utf-8-sig") as file:
    data = file.read()

print("Total number of lines:", len(data.splitlines()))
print("Total number of words:", len(data.split()))
print("Total number of characters:", len(data))
```

```python
# Count characters after removing spaces and newline characters

characters = data.replace(" ", "").replace("\n", "")

print("Characters excluding spaces and newlines:", len(characters))
```

## 10. Searching for a Word

The `input()` function allows the user to enter a word to search for.

The `lower()` method converts text to lowercase so that the search is not affected by capitalization. For example, `Scrooge` and `scrooge` are treated the same.

This example searches for the entered text anywhere in the book.


```python
# Ask the user which word to search for

search_word = input("Enter the word to search: ")

# Search the book without considering capitalization
if search_word.lower() in data.lower():
    print("Word found.")
else:
    print("Word was not found.")
```

```python
# Count the occurrences of a word or phrase

search_word = input("Enter the word or phrase to count: ")

count = data.lower().count(search_word.lower())

print("Occurrences found:", count)
```

## 11. Displaying Lines Containing a Word

Instead of displaying the entire book, we can print only the lines containing a search term.

The `enumerate()` function provides a counter alongside each line. We start counting at `1` so that the displayed line numbers are easier to understand.


```python
# Find and display lines containing a chosen word

search_word = input("Enter a word to find in the book: ")

with open(book_file, "r", encoding="utf-8-sig") as file:
    for line_number, line in enumerate(file, start=1):
        if search_word.lower() in line.lower():
            print("Line", line_number, ":", line.strip())
```

## 12. Copying File Contents

File handling also allows us to copy the contents of one file into another.

First, we read the source file. Next, we open a destination file in `"w"` mode and write the contents into it.

The original book remains unchanged.


```python
# Read the original book
with open(book_file, "r", encoding="utf-8-sig") as source:
    data = source.read()

# Write the contents into a new backup file
with open("christmas_carol_backup.txt", "w", encoding="utf-8") as destination:
    destination.write(data)

print("File copied successfully.")
```

## 13. Saving User Input to a File

The `input()` function returns text, even when the user enters a number.

We can save user-provided information into a file using `write()`. We use `\n` to place each piece of information on a separate line.


```python
# Collect information from the user

name = input("Enter your name: ")
age = input("Enter your age: ")

# Save the information in a text file
with open("user.txt", "w") as file:
    file.write("Name: " + name + "\n")
    file.write("Age: " + age + "\n")

print("Information saved successfully.")
```

## 14. Reading a Specific Number of Characters

The `read()` method can accept a number as an argument. This limits how many characters are read from the file.

For example, `read(10)` reads up to the first ten characters.


```python
# Read only the first 10 characters of the book

with open(book_file, "r", encoding="utf-8-sig") as file:
    first_characters = file.read(10)

print(first_characters)
```
